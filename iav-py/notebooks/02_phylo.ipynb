```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from Bio import Phylo

PHYLO = Path("../results/phylo")
PHYLO
```

```python
tree = Phylo.read(PHYLO / "HA.treefile", "newick")
n_tips = len(tree.get_terminals())

fig, ax = plt.subplots(figsize=(8, max(6, n_tips * 0.12)))
Phylo.draw(
    tree,
    axes=ax,
    do_show=False,
    label_func=lambda c: c.name if c.is_terminal() else "",
)
ax.set_axis_off()
plt.show()
```
