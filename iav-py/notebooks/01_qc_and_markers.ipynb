# 01 QC、分型与功能标记

读取核心流程输出，做快速检查。运行前请先执行 `snakemake -j 8`。

```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt

RESULTS = Path("../results")
RESULTS
```

```python
files = sorted((RESULTS / "consensus").glob("*.stats.tsv"))
stats = (
    pd.concat([pd.read_csv(p, sep="\t") for p in files], ignore_index=True)
    if files else pd.DataFrame()
)
if not stats.empty:
    stats["mean_depth"] = pd.to_numeric(stats["mean_depth"], errors="coerce")
stats
```

```python
if not stats.empty:
    pivot = stats.pivot_table(index="segment", columns="sample", values="mean_depth")
    pivot.plot(kind="bar", figsize=(9, 4))
    plt.ylabel("mean depth")
    plt.tight_layout()
    plt.show()
```

```python
nc_files = sorted((RESULTS / "nextclade").glob("*.tsv"))
nextclade = (
    pd.concat([pd.read_csv(p, sep="\t") for p in nc_files], ignore_index=True)
    if nc_files else pd.DataFrame()
)
cols = ["sample", "segment", "clade", "qc_status", "aa_substitutions"]
nextclade[[c for c in cols if c in nextclade.columns]] if not nextclade.empty else nextclade
```

```python
mk_files = sorted((RESULTS / "markers").glob("*.tsv"))
markers = (
    pd.concat([pd.read_csv(p, sep="\t") for p in mk_files], ignore_index=True)
    if mk_files else pd.DataFrame()
)
markers[markers["found"] == "yes"] if not markers.empty else markers
```
